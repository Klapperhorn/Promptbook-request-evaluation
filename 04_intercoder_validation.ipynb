# 04 Inter-coder validation

Agreement between the manual annotations of several coders (Krippendorff's α per code).

Cells tagged `api` call the LLM (OpenAI or Nebula) and were **not executed** for this repository. `dummy_data/` holds 12 synthetic citizen answers (plus 4 expert rows), random coding runs and random manual codings with the original file names. All results shown are meaningless.

```python
# Folders. The original notebook read and wrote files in the project folder:
# DATA_DIR = OUT_DIR = "."
DATA_DIR = "dummy_data"   # synthetic example data shipped with this repo
OUT_DIR = "output"        # written by the notebook (gitignored)
import os; os.makedirs(OUT_DIR, exist_ok=True)
```

```python
import pandas as pd
# THis is only for the inter-coder validation
# changed: coder initials replaced by Coder_A / Coder_B / Coder_C (variables, column labels, dummy file names)
Annotation=pd.read_excel(f"{DATA_DIR}/MANUAL CODING/2026-04-16 causes Manual Annotation_Coder_A.xlsx", sheet_name="Sheet1",header=0).T
#
Text=Annotation.iloc[1:,-3]
Annotation_A=Annotation.iloc[1:52,:-4]
Annotation_B=pd.read_excel(f"{DATA_DIR}/MANUAL CODING/2026-04-16 causes Manual Annotation_Coder_B.xlsx", sheet_name="Sheet1",header=0).T.iloc[1:52,:-4]
Annotation_C=pd.read_excel(f"{DATA_DIR}/MANUAL CODING/2026-04-16 causes Manual Annotation_Coder_C.xlsx", sheet_name="Sheet1",header=0).T.iloc[1:52,:-4]

Annotation_A.columns=Annotation.iloc[0,:-4]
Annotation_B.columns=Annotation.iloc[0,:-4]
Annotation_C.columns=Annotation.iloc[0,:-4]


filelist=[Annotation_A+Annotation_B+Annotation_C]

Manual_combined = pd.concat(filelist)

Manual_combined=Manual_combined.replace({
    0:True,
    3:True,
    2:False,
    1:False    
})

n_agree=Manual_combined.sum()
s_agree=n_agree/Annotation_C.shape[0]

dis_A_B=Annotation_A.eq(Annotation_B).sum()/Manual_combined.shape[0]
dis_A_C=Annotation_A.eq(Annotation_C).sum()/Manual_combined.shape[0]
dis_B_C=Annotation_C.eq(Annotation_B).sum()/Manual_combined.shape[0]

full_comp=pd.DataFrame({"all_agree [n]":n_agree,"all_agree [%]":s_agree,"eq_A_B":dis_A_B,"eq_A_C":dis_A_C,"eq_B_C":dis_B_C}).sort_values("all_agree [%]").round(2)
print(full_comp.mean())
full_comp
```
